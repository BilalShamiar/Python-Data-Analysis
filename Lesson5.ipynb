```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 150, 120, 200]

plt.figure(figsize=(8,7))

plt.plot(months,sales, color='black', linewidth=3)

plt.title('Montly Sales', fontsize=16, fontweight='bold')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.grid(axis='y', alpha=0.25)
plt.gca().set_facecolor('#f5f7fa')

plt.tight_layout()
plt.savefig("monthly_sales.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python

data = {
    "Student": ["Amy", "Jackson", "Smith", "Oliver", "Tobey",
                "Howard", "Jennifer", "Lisa", "Jeremy", "Natasha"],
    "Math": [78, 92, 65, 88, 72, 95, 81, 69, 85, 90],
    "Python": [85, 96, 70, 91, 68, 89, 94, 75, 82, 98],
    "English": [72, 88, 80, 85, 76, 91, 79, 83, 87, 94],
    "Attendance": [85, 98, 72, 90, 68, 96, 88, 75, 82, 95],
    "Study_Hours": [2, 5, 1, 4, 2, 5, 4, 2, 3, 6]
}

df = pd.DataFrame(data)

df['Average']  = df[['Math','Python','English']].mean(axis=1).round(1)
df
```

```python
plt.figure(figsize=(10,7))

bars = plt.bar(df['Student'],df['Average'], color='#171d38', linewidth=3)

plt.title("Student's Average Marks", fontsize=16, fontweight='bold' )
plt.xlabel('Students')
plt.ylabel('Average Marks')

plt.grid(axis='y', alpha=0.10)
plt.gca().set_facecolor('#f5f7fa')

plt.tight_layout()
plt.bar_label(bars, fmt='%.1f',padding=5)

plt.savefig('students_avg.png', dpi=400, bbox_inches='tight')
plt.show()
```

```python
grades = [
    (df['Average'] >= 90).sum(),
    ((df['Average'] >= 80) & (df['Average'] < 90)).sum(),
    ((df['Average'] >= 70) & (df['Average'] < 80)).sum(),
    (df['Average'] <= 70).sum()
]
labels = ['A (90+)', 'B (80-89)', 'C (70-79)', 'D (Below 70)']
colors = ["#AA5E7A", "#7F5D99", "#22396F", "#202526"]

plt.figure(figsize=(10,8))

plt.pie(
    grades,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    explode=[0.05, 0, 0, 0],
    wedgeprops={"edgecolor": "white", "linewidth": 2},
    textprops={'color':'black','fontsize':'14'}
)

plt.title("Student's Grade Distribution",fontsize=16, fontweight='bold',pad=20)
plt.tight_layout()

plt.savefig('student_grade.png', dpi=600, bbox_inches='tight')
plt.show()
```
